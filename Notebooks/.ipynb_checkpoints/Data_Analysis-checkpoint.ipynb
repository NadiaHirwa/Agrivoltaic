```python
# Import libraries
import os
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
# Get working directory
current_dir = os.getcwd()
# Go one directory up to the root directory
project_root_dir = os.path.dirname(current_dir)
# Define paths to the data folders
data_dir = os.path.join(project_root_dir, 'data')
raw_dir = os.path.join(data_dir, 'raw')
processed_dir = os.path.join(data_dir, 'processed')
# Define paths to results folder
results_dir = os.path.join(project_root_dir, 'results')
# Define paths to docs folder
docs_dir = os.path.join(project_root_dir, 'docs')

# Create directories if they do not exist
os.makedirs(raw_dir, exist_ok = True)
os.makedirs(processed_dir, exist_ok = True)
os.makedirs(results_dir, exist_ok = True)
os.makedirs(docs_dir, exist_ok = True)
```

```python
all_data_filename = os.path.join(processed_dir, "merged.csv")
merged_df = pd.read_csv(all_data_filename)
merged_df.head(10)
```

```python
# Filter temperature columns
temp_cols = [col for col in merged_df.columns if "T (oC)" in col]

# Group by Month and describe temperature columns
temp_stats_by_month = merged_df.groupby("Month")[temp_cols].describe().round(2)

# Display the result
print(temp_stats_by_month)
```

```python
# List all columns
print("All columns in the data:")
print(merged_df.columns.tolist())

# Check for presence of key variable types
key_variables = {
    'solar_irradiation': [col for col in merged_df.columns if 'Irr' in col],
    'temperature': [col for col in merged_df.columns if 'T (oC)' in col],
    'humidity': [col for col in merged_df.columns if 'RH (%)' in col],
    'rainfall': [col for col in merged_df.columns if 'P (mm)' in col],
    'soil_moisture': [col for col in merged_df.columns if 'moisture' in col.lower()],
    'date': [col for col in merged_df.columns if 'DATE' in col.upper()],
    'time': [col for col in merged_df.columns if 'TIME' in col.upper()],
    'month': [col for col in merged_df.columns if 'Month' in col]
}

print("\nKey variable columns found:")
for var, cols in key_variables.items():
    print(f"{var}: {cols if cols else 'Not found'}")
```

```python
months_of_interest = ['May', 'June', 'July', 'August', 'October']
present_months = merged_df['Month'].unique()

print("\nMonths present in your data:")
print(present_months)

missing_months = [m for m in months_of_interest if m not in present_months]
if missing_months:
    print(f"Missing months of interest: {missing_months}")
else:
    print("All months of interest are present.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Prepare a mapping for plot type and variable
irradiation_cols = {
    'Agrivoltaic': ['AG-PV P1/Irr (W/m2)', 'AG-PV P2/Irr (W/m2)', 'AG-PV P3/Irr (W/m2)'],
    'Ground-mounted PV': ['PO-PV/Irr (W/m2)'],
    # Add more if you have them
}
temperature_cols = {
    'Agrivoltaic': ['AG-PV P1/T (oC)', 'AG-PV P2/T (oC)', 'AG-PV P3/T (oC)'],
    'Control': ['AO-SS P1/T (oC)', 'AO-SS P2/T (oC)', 'AO-SS P3/T (oC)', 'AO-TI P2/T (oC)', 'AO-TI P3/T (oC)'],
    'Ground-mounted PV': ['PO-PV/T (oC)'],
    # Add more if you have them
}

# 2. Melt data for irradiation
irradiation_data = []
for plot_type, cols in irradiation_cols.items():
    for col in cols:
        if col in merged_df.columns:
            temp_df = merged_df[[col, 'Month']].copy()
            temp_df = temp_df.rename(columns={col: 'Irradiation'})
            temp_df['Plot Type'] = plot_type
            irradiation_data.append(temp_df)
irradiation_long = pd.concat(irradiation_data, ignore_index=True)

# 3. Melt data for temperature
temperature_data = []
for plot_type, cols in temperature_cols.items():
    for col in cols:
        if col in merged_df.columns:
            temp_df = merged_df[[col, 'Month']].copy()
            temp_df = temp_df.rename(columns={col: 'Temperature'})
            temp_df['Plot Type'] = plot_type
            temperature_data.append(temp_df)
temperature_long = pd.concat(temperature_data, ignore_index=True)

# 4. Boxplot: Solar Irradiation by Plot Type
plt.figure(figsize=(10, 6))
sns.boxplot(x='Plot Type', y='Irradiation', data=irradiation_long)
plt.title('Solar Irradiation by Plot Type')
plt.ylabel('Irradiation (W/m2)')
plt.show()

# 5. Boxplot: Temperature by Plot Type
plt.figure(figsize=(10, 6)) 
sns.boxplot(x='Plot Type', y='Temperature', data=temperature_long)
plt.title('Temperature by Plot Type')
plt.ylabel('Temperature (°C)')
plt.show()

# 6. (Optional) Histograms
plt.figure(figsize=(10, 6))
sns.histplot(data=irradiation_long, x='Irradiation', hue='Plot Type', element='step', stat='density', common_norm=False)
plt.title('Distribution of Solar Irradiation')
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(data=temperature_long, x='Temperature', hue='Plot Type', element='step', stat='density', common_norm=False)
plt.title('Distribution of Temperature')
plt.show()
```

```python
# 1. Prepare mapping for plot type and variable
humidity_cols = {
    'Agrivoltaic': [
        'AG-SS P1/RH (%)', 'AG-SS P2/RH (%)', 'AG-SS P3/RH (%)',
        'AG-TI P1/RH (%)', 'AG-TI P2/RH (%)', 'AG-TI P3/RH (%)'
    ],
    'Control': [
        'AO-SS P1/RH (%)', 'AO-SS P2/RH (%)', 'AO-SS P3/RH (%)',
        'AO-TI P2/RH (%)', 'AO-TI P3/RH (%)'
    ]
}

# 2. Melt data for humidity, keeping DATE
humidity_data = []
for plot_type, cols in humidity_cols.items():
    for col in cols:
        if col in merged_df.columns:
            temp_df = merged_df[[col, 'Month', 'DATE']].copy()
            temp_df = temp_df.rename(columns={col: 'Humidity'})
            temp_df['Plot Type'] = plot_type
            humidity_data.append(temp_df)
humidity_long = pd.concat(humidity_data, ignore_index=True)

# Now you can safely convert the DATE column
humidity_long['Date'] = pd.to_datetime(humidity_long['DATE'])
humidity_daily = humidity_long.groupby(['Date', 'Plot Type'])['Humidity'].mean().reset_index()

# Plot
plt.figure(figsize=(14, 6))
sns.lineplot(x='Date', y='Humidity', hue='Plot Type', data=humidity_daily)
plt.title('Daily Mean Humidity by Plot Type')
plt.ylabel('Humidity (%)')
plt.show()

# 3. Rainfall (assuming one column for all plots)
rainfall_long = merged_df[['WS/P (mm)', 'Month']].copy()
rainfall_long = rainfall_long.rename(columns={'WS/P (mm)': 'Rainfall'})
rainfall_long['Plot Type'] = 'All'  # If you have rainfall by plot, adjust accordingly

# 4. Boxplot: Humidity by Plot Type
plt.figure(figsize=(10, 6))
sns.boxplot(x='Plot Type', y='Humidity', data=humidity_long)
plt.title('Relative Humidity by Plot Type')
plt.ylabel('Humidity (%)')
plt.show()

# 5. Violin plot: Humidity by Plot Type
plt.figure(figsize=(10, 6))
sns.violinplot(x='Plot Type', y='Humidity', data=humidity_long, inner='quartile')
plt.title('Relative Humidity by Plot Type (Violin Plot)')
plt.ylabel('Humidity (%)')
plt.show()

# 6. Boxplot: Rainfall
plt.figure(figsize=(8, 6))
sns.boxplot(x='Plot Type', y='Rainfall', data=rainfall_long)
plt.title('Rainfall Distribution')
plt.ylabel('Rainfall (mm)')
plt.show()

# 7. Time Series: Humidity (mean by day)
humidity_long['Date'] = pd.to_datetime(merged_df['DATE'].values[:len(humidity_long)])
humidity_daily = humidity_long.groupby(['Date', 'Plot Type'])['Humidity'].mean().reset_index()
plt.figure(figsize=(14, 6))
sns.lineplot(x='Date', y='Humidity', hue='Plot Type', data=humidity_daily)
plt.title('Daily Mean Humidity by Plot Type')
plt.ylabel('Humidity (%)')
plt.show()

# 8. Time Series: Rainfall (sum by day)
rainfall_long['Date'] = pd.to_datetime(merged_df['DATE'])
rainfall_daily = rainfall_long.groupby('Date')['Rainfall'].sum().reset_index()
plt.figure(figsize=(14, 6))
sns.lineplot(x='Date', y='Rainfall', data=rainfall_daily)
plt.title('Daily Rainfall')
plt.ylabel('Rainfall (mm)')
plt.show()
```
